# 第 3 章 · 抽样与中心极限定理

现实中我们无法测量整个**总体**，只能抽取**样本**去推断总体。
抽样分布和中心极限定理（CLT）是这种推断的地基。

## 0. 准备环境

```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 让 matplotlib 正确显示中文与负号
plt.rcParams["font.sans-serif"] = ["Microsoft YaHei", "SimHei"]
plt.rcParams["axes.unicode_minus"] = False

np.random.seed(42)   # 固定随机种子，保证每次运行结果一致
print("环境就绪：numpy", np.__version__, "| pandas", pd.__version__)
```

## 1. 总体与样本

- **总体**：研究对象全体，参数常用 μ（均值）、σ（标准差）表示。
- **样本**：从总体中抽取的一部分，统计量用 x̄（样本均值）、s 表示。
- **抽样误差**：样本统计量与总体参数的差异，无法完全消除，但可以估计。

```python
# 构造一个"总体"：10000 名学生的成绩（右偏，不是正态）
population = np.random.gamma(shape=5, scale=12, size=10000) + 20
population = population[:10000]

print("总体大小 =", len(population))
print("总体均值 μ = %.2f" % population.mean())
print("总体标准差 σ = %.2f" % population.std(ddof=0))

plt.hist(population, bins=40, color="#b9c8dd", edgecolor="white")
plt.title("总体分布（明显右偏，不是正态）")
plt.xlabel("成绩"); plt.show()
```

## 2. 抽样分布

对同一总体反复抽样，每次算一个样本均值 x̄，这些 x̄ 构成的分布叫**抽样分布**。

- E(x̄) = μ  （样本均值是总体均值的无偏估计）
- 标准误 SE = σ / √n  （样本均值的标准差，衡量估计的精度）

```python
n = 30                      # 每次抽 30 人
means = [np.random.choice(population, n, replace=False).mean() for _ in range(2000)]

print("抽样分布均值 = %.2f  (接近总体均值 %.2f)" % (np.mean(means), population.mean()))
print("抽样分布标准差(标准误) = %.3f" % np.std(means, ddof=1))
print("理论标准误 σ/√n = %.3f" % (population.std() / math.sqrt(n)))
```

## 3. 中心极限定理（CLT）

**核心结论**：无论总体是什么分布，只要样本量 n 足够大，
样本均值 x̄ 的分布就近似 **正态分布** N(μ, σ²/n)。

下面用刚才那个右偏总体做模拟，肉眼可见地验证 CLT。

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
for ax, n in zip(axes.ravel(), [1, 5, 30, 100]):
    means = [np.random.choice(population, n, replace=False).mean() for _ in range(2000)]
    ax.hist(means, bins=35, color="#6da8f5", edgecolor="white", density=True)
    ax.set_title(f"样本量 n = {n}")
    ax.set_xlabel("样本均值")
plt.suptitle("随着 n 增大，样本均值分布越来越接近正态（CLT）", fontsize=13)
plt.tight_layout()
plt.show()
```

## 4. 标准误随样本量变化

标准误 SE = σ/√n：**样本量翻 4 倍，误差才减半**——这就是"精度要花钱买"的数学表达。

```python
ns = [5, 10, 20, 50, 100, 200, 500]
se = [population.std() / math.sqrt(k) for k in ns]
for k, s in zip(ns, se):
    print(f"n = {k:>3}  ->  SE = {s:.3f}")

plt.figure(figsize=(7, 4))
plt.plot(ns, se, "o-", color="#2255aa")
plt.xlabel("样本量 n"); plt.ylabel("标准误 SE")
plt.title("标准误与样本量的关系"); plt.show()
```

## 小结

- 统计推断 = 用样本统计量估计总体参数。
- 抽样分布描述"统计量本身"的波动；**标准误**是它的标准差。
- **中心极限定理**让正态分布成为一切推断的通用工具（哪怕总体不正态）。
